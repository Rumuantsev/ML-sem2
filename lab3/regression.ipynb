Можно увидеть, что зависимость между x и y выражается как y = −x. То есть оптимальные 
значения весов сети будет w1 = −1, w0 = 0.  
Напишите программу для обнаружения этих значений при помощи алгоритма 
классического ГС. Для этого вам потребуется составить функцию потерь L(w). Частные 
производные можете посчитать аналитически и использовать в программе. Для старта 
спуска возьмем a0 = (0,0), h = 0.1, причем мы предполагаем, что первая координата позиции 
"шарика" соответствует весу w0, а вторая координата соответствует весу w1. 
Сколько итераций потребовалось, чтобы достигнуть точки минимума?

```python
import math

xs = [-1, 0, 1]
ys = [ 1, 0,-1]

h = 0.1           
w0 = 0.0         
w1 = 0.0        
tol = 1e-6    
max_iter = 10000

def loss(w0,w1):
    return sum((w1*x + w0 - y)**2 for x,y in zip(xs,ys))

def grad(w0,w1):
    d_w0 = 6.0 * w0
    d_w1 = 4.0 * (w1 + 1.0)
    return d_w0, d_w1

target = (0.0, -1.0)
def dist(w0,w1):
    return math.hypot(w0 - target[0], w1 - target[1])

it = 0
while dist(w0,w1) > tol and it < max_iter:
    d0, d1 = grad(w0,w1)
    w0 -= h * d0
    w1 -= h * d1
    it += 1

print("Итераций:", it)
print("w0 =", w0, "w1 =", w1)
print("Loss =", loss(w0,w1))
```

программе. Для старта спуска возьмем a0 = (0,0), h = 0.1, причем мы предполагаем, что 
первая координата позиции "шарика" соответствует весу w0, а вторая координата

```python
import math

xs = [0,1,2]
ys = [1,2,3]

h = 0.1          
w0 = 0.0          
w1 = 0.0          
tol = 1e-6        
max_iter = 10000

def loss(w0,w1):
    return sum((w1*x + w0 - y)**2 for x,y in zip(xs,ys))

def grad(w0,w1):
    d_w0 = 6.0 * (w0 + w1 - 2.0)
    d_w1 = 10.0 * w1 + 6.0 * w0 - 16.0
    return d_w0, d_w1

target = (1.0, 1.0)
def dist(w0,w1):
    return math.hypot(w0 - target[0], w1 - target[1])

it = 0
while dist(w0,w1) > tol and it < max_iter:
    d0, d1 = grad(w0,w1)
    w0 -= h * d0
    w1 -= h * d1
    it += 1

print("Итераций:", it)
print("w0 =", w0, "w1 =", w1)
print("Loss =", loss(w0,w1))
```

Теперь измените программу из задания №2 и замените классический ГС на стохастический. 
Так как в ТВ у нас 3 объекта, то на каждой эпохе у вас будет три итерации. При этом за 
одну эпоху должны быть использованы все объекты, но в случайном порядке. 
Сколько итераций потребовалось, чтобы достигнуть точки минимума?

```python
import random, math

xs = [0,1,2]
ys = [1,2,3]
h = 0.1
tol = 1e-6
max_iters = 100000

w0, w1 = 0.0, 0.0
it = 0
rnd = random.Random(42)   

def dist2(w0, w1, target=(1.0,1.0)):
    dx = w0 - target[0]
    dy = w1 - target[1]
    return math.sqrt(dx*dx + dy*dy)

while dist2(w0, w1) > tol and it < max_iters:
    idxs = [0,1,2]
    rnd.shuffle(idxs)          
    for i in idxs:
        x, y = xs[i], ys[i]
        e = (w1*x + w0 - y)
        d0 = 2*e
        d1 = 2*e*x
        w0 -= h * d0
        w1 -= h * d1
        it += 1
        if dist2(w0, w1) <= tol:
            break

print("Итераций:", it)
print("Эпох (прибл.):", it/3.0)
print("w0, w1:", w0, w1)
```

Теперь измените программу из задания №1 и замените классический ГС на стохастический 
по мини-батчам. Размер батча равен 2. Так как в ТВ у нас 3 объекта, то на каждой эпохе у 
вас будет две итерации. При этом батчи у вас будут пересекаться. Например, в первом батче 
будут 1 и 2 объекты, а во втором батче – 2 и 3 объекты. Главное, чтобы за одну эпоху были 
использованы все объекты. 
Сколько итераций потребовалось, чтобы достигнуть точки минимума? 

```python
import math

xs = [-1, 0, 1]
ys = [ 1,  0,-1]

h = 0.1
w0 = 0.0
w1 = 0.0
tol = 1e-6
max_iters = 100000

batches = [(0,1), (1,2)]

def grad_single(x, y, w0, w1):
    e = (w1 * x + w0 - y)
    return 2*e, 2*e*x

def close_enough(w0, w1, target=(0.0, -1.0), tol=1e-6):
    dx = w0 - target[0]
    dy = w1 - target[1]
    return dx*dx + dy*dy <= tol*tol

it = 0
while not close_enough(w0, w1) and it < max_iters:
    for batch in batches:

        d0_sum = 0.0
        d1_sum = 0.0
        for i in batch:
            d0_i, d1_i = grad_single(xs[i], ys[i], w0, w1)
            d0_sum += d0_i
            d1_sum += d1_i

        k = len(batch)
        d0_avg = d0_sum / k
        d1_avg = d1_sum / k

        w0 -= h * d0_avg
        w1 -= h * d1_avg

        it += 1
        if close_enough(w0, w1):
            break

print("Итераций (батч-апдейты):", it)
print("w0, w1:", w0, w1)
print("Loss:", sum((w1*x + w0 - y)**2 for x,y in zip(xs,ys)))
```

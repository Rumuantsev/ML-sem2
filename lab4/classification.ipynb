##### 2  
$
-(\ln(0.6) + \ln(x)) = 2\ln(2)
$

$
-\ln(0.6) - \ln(x) = \ln(4)
$

$

$

$
-\ln(x) = \ln(2.4)
$

$
\ln(x) = \ln\left(\frac{1}{2.4}\right)
$

$

##### 3
$
-(\ln(0.6) + \ln(x)) = \frac{2\ln(2)}{1.5}
$

$
\frac{2\ln(2)}{1.5} = \frac{4}{3}\ln(2) = \ln(2^{4/3})
$

$
-(\ln(0.6) + \ln(x)) = -\ln(0.6) - \ln(x)
$

$
-\ln(x) = \ln(2^{4/3}) + \ln(0.6)
$

$
\ln(2^{4/3}) + \ln(0.6) = \ln(2^{4/3} \times 0.6) = \ln\left(2^{4/3} \times \frac{3}{5}\right)
$

$
x = \frac{5}{3 \times 2^{4/3}} = \frac{5}{6 \times \sqrt[3]{2}}
$

$
x = \frac{5}{6\sqrt[3]{2}} \approx 0.6612
$

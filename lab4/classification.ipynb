Рассмотрим два дурацких классификатора К1 и К2.

Дурацкий классификатор К2 был натренирован по тренировочной выборке, содержащей 
10 объектов класса 0 и 10 объектов класса 1. 

Оба классификатора выдают одинаковые ответы (они для любого объекта говорят, что он с

Однако значения функции потерь на тренировочной выборке у них различны (вот это 
поворот!). 

Найдите значение функции потерь для каждого классификатора.

выборки классификатора К2. Это означает, что К2 "более глуп", чем К1. Дело в том, что К2 
имел в своем распоряжении бОльшую выборку, чем К1, но ничего кроме дурацкой
